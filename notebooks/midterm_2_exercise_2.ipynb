```python
x, y, z = sp.symbols('x, y, z')
s1, c1, s12, c12 = sp.symbols('s1, c1, s12, c12')
```

```python
equations = [
    a2*c1 + a4*c12 + (a5+d3)*s12 - x,
    a2*s1 + a4*s12 - (a5+d3)*c12 - y,
    a1 + a3 - z,
    s1**2 + c1**2 - 1,
    s12**2 + c12**2 - 1
]
solutions = sp.nonlinsolve(equations, [s1, c1, s12, c12, d3])
solutions
# solutions ke vrati lista od torki za site razlicni mozni resenija za sin1, sin1+2, cos1, cos1+2 i d3 izrazeni preku promenlivite
# x, y i z koi se definirani vo linija [5] 
```
