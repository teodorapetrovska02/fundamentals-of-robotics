## Zadacha 2

```python
m1, m2, l1, l2, g = sp.symbols('m1, m2, l1, l2, g')
theta1, theta2 = mechanics.dynamicsymbols('theta1, theta2')
robot = rt.SerialLinkRobot()
robot.add_revolute_joint(theta1, 0, l1, 0)
robot.add_revolute_joint(theta2, 0, l2, 0)
robot.add_subs([(l1, 1), (l2, 1)])
robot.interact()
```

```python
p1_1 = robot.get_dh_joint_to_joint(0, 0)[:3, 3]
p1_2 = robot.get_dh_joint_to_joint(0, 1)[:3, 3]
p1 = p1_1 + (p1_2 - p1_1) / 2
p1
```

```python
v1 = p1.diff(t)
v1_squared = v1.dot(v1).simplify()
K1 = m1 * v1_squared / 2
K1
```

```python
p2_1 = robot.get_dh_joint_to_joint(0, 1)[:3, 3]
p2_2 = robot.get_dh_joint_to_joint(0, 2)[:3, 3]
p2 = p2_1 + (p2_2 - p2_1) / 2
p2
```

```python
v2 = p2.diff(t)
v2_squared = v2.dot(v2).simplify()
K2 = m2 * v2_squared / 2
K2
```

```python
K = K1 + K2
K
```

```python
P1 = m1 * g * p1[1]
P2 = m2 * g * p2[1]
P = P1 + P2
P
```

```python
L = K - P
L
```

```python
rt.lagrangian(L, [theta1, theta2])
```

```python
rt.dynamic_model_with_3_matrices(L, [theta1, theta2])
```
